```python
import pandas as pd
import re
import os
import json
```

```python
directory = ['Algorithm','Data_structure']
```

```python
#планируется словарь алгоритмов для более быстрого создания всех датафреймов
alg_dict = {'sort': ['bubble_sort', 'cocktail_sort','comb_sort','counting_sort','gnome_sort','heap_sort',
                     'insertion_sort','merge_sort','quick_sort','radix_sort','selection_sort','---'],
            'find': ['binary_search', 'linear_search', 'quick_select']}
```

```python
data_arr = ['CyclicNode', 'DoubleLinkedList','DynamicArray','HashTable','SingleLinkedList','AVL_Tree',
            'Binary_Tree','Heap','Queue','Set','Trie']
```

```python
def natural_sort_key(s):
    return [int(t) if t.isdigit() else t for t in re.split(r'(\d+)', s)]

def remove_chars(string):
    chars_to_remove = '()0123456789 '
    return ''.join(char for char in string if char not in chars_to_remove)

def make_dataset_alg(directory,alg_dict,method):
    df = pd.DataFrame(columns = ['output','input','instruction'])
    for key, values in alg_dict.items():
        for value in values:
            file_path = f'dataset\{directory}\{key}\{value}\{method}'
            file_list = sorted(os.listdir(file_path), key=natural_sort_key)
            for file_name in file_list:
                with open(file_path + '/' + file_name,'r', encoding = 'utf-8') as f:
                    lines = f.readlines()
                file_alg = remove_chars(file_name)
                if file_name[-2:] != ').py':
                    df.loc[len(df)] = [file_alg[:-4] ,'',''.join(lines)]
    return df

def make_dataset_data(directory,data_arr):
    df = pd.DataFrame(columns = ['output','input','instruction'])
    for i in data_arr:
        file_path = f'dataset\{directory}\{i}'
        file_list = sorted(os.listdir(file_path), key=natural_sort_key)
        for file_name in file_list:
            with open(file_path + '/' + file_name,'r', encoding = 'utf-8') as f:
                lines = f.readlines()
            file_data = remove_chars(file_name)
            if file_name[-2:] != ').py':
                df.loc[len(df)] = [file_data[:-4] ,'',''.join(lines)]
    return df

def create_file(name,method = None):
    if method:
        df = make_dataset_alg(directory[0],alg_dict,method)
        df.to_csv(f'csv_for_LLaMa\{name}.csv')
        data = df.to_dict('records')
    
        with open(f'csv_for_LLaMa/{name}.json', 'w', encoding='utf-8') as f:
            json.dump(data, f, indent=2)
    else:
        df = make_dataset_data(directory[1],data_arr)
        df.to_csv(f'csv_for_LLaMa\{name}.csv')
        data = df.to_dict('records')
    
        with open(f'csv_for_LLaMa/{name}.json', 'w', encoding='utf-8') as f:
            json.dump(data, f, indent=2)
    return df
    
```
