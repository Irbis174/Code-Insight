```python
import pandas as pd
import re
import os
import json
```

```python
directory = ['Algorithm','Data_structure']
```

```python
#планируется словарь алгоритмов для более быстрого создания всех датафреймов
alg_dict = {'sort': ['bubble_sort', 'quick_sort','selection_sort','merge_sort','insertion_sort'], 'find': ['binary_search', 'linear_search', 'quick_select']}
```

```python
data_arr = ['CyclicNode', 'DoubleLinkedList','DynamicArray','HashTable','SingleLinkedList']
```

```python
def natural_sort_key(s):
    return [int(t) if t.isdigit() else t for t in re.split(r'(\d+)', s)]

def remove_chars(string):
    chars_to_remove = '()0123456789'
    return ''.join(char for char in string if char not in chars_to_remove)

def make_dataset_alg(directory,alg_dict):
    df = pd.DataFrame(columns = ['output','input','instruction'])
    for key, values in alg_dict.items():
        for value in values:
            file_path = f'dataset\{directory}\{key}\{value}'
            file_list = sorted(os.listdir(file_path), key=natural_sort_key)
            for file_name in file_list:
                with open(file_path + '/' + file_name,'r', encoding = 'utf-8') as f:
                    lines = f.readlines()
                file_alg = remove_chars(file_name)
                if file_name[-2:] != ').py':
                    df.loc[len(df)] = [file_alg[:-3] ,'',''.join(lines)]
    return df

def make_dataset_data(directory,data_arr):
    df = pd.DataFrame(columns = ['output','input','instruction'])
    for i in data_arr:
        file_path = f'dataset\{directory}\{i}'
        file_list = sorted(os.listdir(file_path), key=natural_sort_key)
        for file_name in file_list:
            with open(file_path + '/' + file_name,'r', encoding = 'utf-8') as f:
                lines = f.readlines()
            file_data = remove_chars(file_name)
            if file_name[-2:] != ').py':
                df.loc[len(df)] = [file_data[:-3] ,'',''.join(lines)]
    return df
```

```python
df = make_dataset_data(directory[1],data_arr)
df
```

```python
df.to_csv('csv_for_LLaMa\data.csv')
```

```python
data = df.to_dict('records')

with open('csv_for_LLaMa/data.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=2)
```
